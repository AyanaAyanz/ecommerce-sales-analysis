```python
# ==========================================
# E-Commerce Sales Analysis
# Cell 1: Import Libraries
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully!")

# ==========================================
# Cell 2: Load E-Commerce Dataset
# ==========================================

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"

df = pd.read_excel(url)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())

# ==========================================
# Cell 3: Understand the Dataset
# ==========================================

print("========== DATASET INFORMATION ==========")

# Dataset shape
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# First 5 rows
print("\n========== FIRST 5 ROWS ==========")
display(df.head())

# Last 5 rows
print("\n========== LAST 5 ROWS ==========")
display(df.tail())

# Column names
print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

# Dataset information
print("\n========== DATASET INFO ==========")
df.info()

# Missing values
print("\n========== MISSING VALUES ==========")
display(df.isnull().sum())

# Statistical summary
print("\n========== STATISTICAL SUMMARY ==========")
display(df.describe(include="all"))

# ==========================================
# Cell 5: Data Transformation
# ==========================================

# Calculate total sales for each transaction
df["Sales"] = df["Quantity"] * df["UnitPrice"]

# Extract date-related features
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Month_Name"] = df["InvoiceDate"].dt.month_name()
df["Day"] = df["InvoiceDate"].dt.day
df["Day_Name"] = df["InvoiceDate"].dt.day_name()
df["Hour"] = df["InvoiceDate"].dt.hour

# Create a Year-Month column for monthly sales analysis
df["Year_Month"] = df["InvoiceDate"].dt.to_period("M").astype(str)

print("========== DATA TRANSFORMATION COMPLETED ==========")

print("\nNew columns created:")
print([
    "Sales",
    "Year",
    "Month",
    "Month_Name",
    "Day",
    "Day_Name",
    "Hour",
    "Year_Month"
])

print("\nUpdated dataset:")
display(df.head())

print("\nUpdated dataset shape:", df.shape)

# ==========================================
# Cell 6: Basic E-Commerce Sales Analysis
# ==========================================

# Total Revenue
total_revenue = df["Sales"].sum()

# Total Orders
total_orders = df["InvoiceNo"].nunique()

# Total Products
total_products = df["StockCode"].nunique()

# Total Customers
total_customers = df["CustomerID"].nunique()

# Total Countries
total_countries = df["Country"].nunique()

# Total Items Sold
total_items = df["Quantity"].sum()

print("========== E-COMMERCE SALES SUMMARY ==========")

print(f"Total Revenue     : £{total_revenue:,.2f}")
print(f"Total Orders      : {total_orders:,}")
print(f"Total Products    : {total_products:,}")
print(f"Total Customers   : {total_customers:,}")
print(f"Total Countries   : {total_countries:,}")
print(f"Total Items Sold  : {total_items:,}")

# Average Order Value
average_order_value = total_revenue / total_orders

print(f"Average Order Value: £{average_order_value:,.2f}")
# ==========================================
# Cell 8: Top-Selling Products
# ==========================================

# Group products by description and calculate total quantity sold
top_products = (
    df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

print("========== TOP 10 BEST-SELLING PRODUCTS ==========")
display(top_products)

# ------------------------------------------
# Visualization
# ------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_products,
    x="Quantity",
    y="Description"
)

plt.title("Top 10 Best-Selling Products by Quantity")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
# ==========================================
# Cell 9: Product Revenue Performance
# ==========================================

# Calculate total revenue for each product
product_revenue = (
    df.groupby("Description")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

print("========== TOP 10 PRODUCTS BY REVENUE ==========")
display(product_revenue)

# ------------------------------------------
# Visualization
# ------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=product_revenue,
    x="Sales",
    y="Description"
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Revenue (£)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()

# ==========================================
# Cell 9: Product Revenue Performance
# ==========================================

# Calculate total revenue for each product
product_revenue = (
    df.groupby("Description")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

print("========== TOP 10 PRODUCTS BY REVENUE ==========")
display(product_revenue)

# ------------------------------------------
# Visualization
# ------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=product_revenue,
    x="Sales",
    y="Description"
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Revenue (£)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()

# ==========================================
# Cell 10: Sales by Country / Region
# ==========================================

# Calculate total sales by country
country_sales = (
    df.groupby("Country")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

print("========== TOP 10 COUNTRIES BY SALES ==========")
display(country_sales)

# ------------------------------------------
# Find the highest-sales country
# ------------------------------------------

top_country = country_sales.iloc[0]

print("\n========== TOP SALES COUNTRY ==========")
print("Country:", top_country["Country"])
print(f"Sales: £{top_country['Sales']:,.2f}")

# ------------------------------------------
# Visualization
# ------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=country_sales,
    x="Sales",
    y="Country"
)

plt.title("Top 10 Countries by Sales")
plt.xlabel("Total Sales (£)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

# ==========================================
# Cell 11: Customer & Order Analysis
# ==========================================

# ------------------------------------------
# 1. Revenue generated by each customer
# ------------------------------------------

customer_sales = (
    df.dropna(subset=["CustomerID"])
      .groupby("CustomerID")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

print("========== TOP 10 CUSTOMERS BY REVENUE ==========")
display(customer_sales)

# ------------------------------------------
# 2. Number of orders per customer
# ------------------------------------------

customer_orders = (
    df.dropna(subset=["CustomerID"])
      .groupby("CustomerID")["InvoiceNo"]
      .nunique()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name="Number_of_Orders")
)

print("\n========== TOP 10 CUSTOMERS BY NUMBER OF ORDERS ==========")
display(customer_orders)

# ------------------------------------------
# 3. Average Order Value
# ------------------------------------------

average_order_value = df.groupby("InvoiceNo")["Sales"].sum().mean()

print("\n========== ORDER ANALYSIS ==========")
print(f"Average Order Value: £{average_order_value:,.2f}")

# ------------------------------------------
# 4. Visualization - Top Customers
# ------------------------------------------

plt.figure(figsize=(10, 6))

sns.barplot(
    data=customer_sales,
    x="Sales",
    y="CustomerID"
)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Total Revenue (£)")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()

# ==========================================
# Cell 12: Sales by Day & Time
# ==========================================

# ------------------------------------------
# 1. Sales by Day of the Week
# ------------------------------------------

day_sales = (
    df.groupby("Day_Name")["Sales"]
    .sum()
    .reindex([
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday",
        "Sunday"
    ])
)

print("========== SALES BY DAY OF THE WEEK ==========")
display(day_sales.reset_index(name="Sales"))

# Find highest-sales day
highest_day = day_sales.idxmax()

print("Highest Sales Day:", highest_day)
print(f"Sales: £{day_sales.max():,.2f}")

# ------------------------------------------
# Visualization: Sales by Day
# ------------------------------------------

plt.figure(figsize=(10, 5))

sns.barplot(
    x=day_sales.index,
    y=day_sales.values
)

plt.title("Sales by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Total Sales (£)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------
# 2. Sales by Hour
# ------------------------------------------

hour_sales = (
    df.groupby("Hour")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n========== SALES BY HOUR ==========")
display(hour_sales.reset_index(name="Sales"))

# Find highest-sales hour
highest_hour = hour_sales.idxmax()

print("Highest Sales Hour:", highest_hour)
print(f"Sales: £{hour_sales.max():,.2f}")

# ------------------------------------------
# Visualization: Sales by Hour
# ------------------------------------------

plt.figure(figsize=(10, 5))

sns.lineplot(
    x=hour_sales.index,
    y=hour_sales.values,
    marker="o"
)

plt.title("Sales by Hour of the Day")
plt.xlabel("Hour")
plt.ylabel("Total Sales (£)")
plt.xticks(range(0, 24))
plt.grid(True)
plt.tight_layout()
plt.show()


# ==========================================
# Cell 13: Final Business Insights
# ==========================================

print("================================================")
print("        E-COMMERCE SALES ANALYSIS")
print("        FINAL BUSINESS INSIGHTS")
print("================================================")

# ------------------------------------------
# Key Business Metrics
# ------------------------------------------

print("\n========== KEY BUSINESS METRICS ==========")
print(f"Total Revenue       : £{total_revenue:,.2f}")
print(f"Total Orders        : {total_orders:,}")
print(f"Total Products      : {total_products:,}")
print(f"Total Customers     : {total_customers:,}")
print(f"Total Countries     : {total_countries:,}")
print(f"Total Items Sold    : {total_items:,}")
print(f"Average Order Value : £{average_order_value:,.2f}")

# ------------------------------------------
# Top Product by Quantity
# ------------------------------------------

top_quantity_product = top_products.iloc[0]

print("\n========== TOP-SELLING PRODUCT ==========")
print("Product:", top_quantity_product["Description"])
print(f"Quantity Sold: {top_quantity_product['Quantity']:,.0f}")

# ------------------------------------------
# Top Product by Revenue
# ------------------------------------------

top_revenue_product = product_revenue.iloc[0]

print("\n========== TOP REVENUE PRODUCT ==========")
print("Product:", top_revenue_product["Description"])
print(f"Revenue: £{top_revenue_product['Sales']:,.2f}")

# ------------------------------------------
# Highest Sales Month
# ------------------------------------------

monthly_sales = (
    df.groupby("Year_Month")["Sales"]
    .sum()
    .reset_index()
)

highest_sales_month = monthly_sales.loc[
    monthly_sales["Sales"].idxmax()
]

print("========== HIGHEST SALES MONTH ==========")
print("Month:", highest_sales_month["Year_Month"])
print(f"Sales: £{highest_sales_month['Sales']:,.2f}")

# ------------------------------------------
# Top Country
# ------------------------------------------

print("\n========== TOP COUNTRY ==========")
print("Country:", top_country["Country"])
print(f"Sales: £{top_country['Sales']:,.2f}")

# ------------------------------------------
# Highest Sales Day
# ------------------------------------------

print("\n========== HIGHEST SALES DAY ==========")
print("Day:", highest_day)
print(f"Sales: £{day_sales.max():,.2f}")

# ------------------------------------------
# Highest Sales Hour
# ------------------------------------------

print("\n========== HIGHEST SALES HOUR ==========")
print(f"Hour: {highest_hour}:00")
print(f"Sales: £{hour_sales.max():,.2f}")

# ------------------------------------------
# Business Recommendations
# ------------------------------------------

print("\n========== BUSINESS RECOMMENDATIONS ==========")

print("1. Focus inventory on high-demand products.")
print("2. Promote high-revenue products to increase profitability.")
print("3. Target the strongest countries with localized marketing.")
print("4. Run promotions during high-performing sales periods.")
print("5. Use customer purchase behavior for targeted marketing.")
print("6. Monitor monthly sales trends to improve inventory planning.")

print("\n================================================")
print("          ANALYSIS COMPLETED SUCCESSFULLY")
print("================================================")
```
